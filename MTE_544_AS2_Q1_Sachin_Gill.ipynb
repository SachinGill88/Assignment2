```python
## import libraries, just need numpy
import numpy as np 
```

```python
## define the points in the question, so we can create an exact matrix to solve for the coefficients
points = np.array([
    [2.92, -6.01],
    [3.40, -7.20],
    [4.99, -7.84],
    [5.48, -7.04],
    [4.20, -5.91],
])
x, y = points.T
M = np.column_stack((x**2, 2*x*y, y**2, -2*x, -2*y))
X = np.array([-1,-1,-1,-1,-1])

A,B,C,D,E = np.linalg.solve (M, X)

## Solve for the coeffecients (A,B,C,D,E)
print (A,B,C,D,E)
```

```python
## Solve for the other values (Q,b) using the relationships given
# Define matrix F 
F = np.array([[A, B], [B, C]])
G = np.array([D, E])


alpha = 1 / (G @ np.linalg.inv(F) @ G - 1)
Q = alpha * F
b = np.linalg.solve(Q, alpha * G)

## print the results
print("alpha:",alpha)
print("Q:",Q)
print("b:", b)
```
