```python
import numpy as np 
```

```python
M = np.array([
    [ 8.53, -35.10, 36.12,  -5.84, 12.02],
    [11.56, -48.96, 51.84,  -6.80, 14.40],
    [24.90, -78.24, 61.47,  -9.98, 15.68],
    [30.03, -77.16, 49.56, -10.96, 14.08],
    [17.64, -49.64, 34.93,  -8.40, 11.82],
])

X = np.array([-1,-1,-1,-1,-1])

A,B,C,D,E = np.linalg.solve (M, X)

## Solve for the coeffecients (A,B,C,D,E)
print (A,B,C,D,E)
```

```python
## Solve for the other values (Q,b) using the relationships given
# Define matrix F 
F = np.array([[A, B], [B, C]])
G = np.array([D, E])


alpha = 1 / (G @ np.linalg.inv(F) @ G - 1)
Q = alpha * F
b = np.linalg.solve(Q, alpha * G)

print("alpha:",alpha)
print("Q:",Q)
print("b:", b)
```
